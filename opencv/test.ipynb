```python
import numpy as np
import os

ls = os.listdir('data')
filename = "data/"+ls[4]

# 1. .bin 파일에서 데이터 로드
loaded_data = np.fromfile(filename, dtype=np.uint8)  # uint8 형식으로 데이터 로드

# 2. 데이터 크기 복원 (4, 224, 224, 3) - 저장 시의 차원
loaded_images = loaded_data.reshape((4, 196, 768))

# 3. 결과 출력
print("로드된 데이터의 shape:", loaded_images.shape)  # (4, 224, 224, 3)
```

```python
imgs = np.transpose(loaded_images.reshape(4, 14, 14, 3, 16, 16), (0, 1, 4, 2, 5, 3)).reshape(4, 224, 224, 3)
```

```python
img = imgs[3]
if img is None:
    print("이미지를 읽을 수 없습니다. 경로를 확인하세요.")
else:
    img_rgb = img
    # img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    plt.imshow(img_rgb)
    plt.axis('off')
    plt.show()
```

```python
with open('data/floated_img.bin', "wb") as f:
    norm_img = loaded_images.astype(np.float32)/255.
    norm_img.tofile(f)
```
